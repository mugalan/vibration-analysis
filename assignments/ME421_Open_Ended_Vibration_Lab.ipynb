## Week 1 — Mathematical modelling and experimental planning

### Activity 1 — Mathematical formulation

As a starting point, the small-amplitude transverse vibrations of the beam under constant axial tension can be modelled using the Euler–Bernoulli beam equation

$$
\rho A\frac{\partial^2 y}{\partial t^2}
+EI\frac{\partial^4 y}{\partial x^4}
-T\frac{\partial^2 y}{\partial x^2}=f(x,t).
$$

Here $y(x,t)$ is the transverse displacement, $x\in[0,L]$ is the position along the beam, $L$ is the beam length, $\rho$ is the material density, $A$ is the cross-sectional area, $E$ is Young's modulus, $I$ is the second moment of area, $T$ is the axial tension and $f(x,t)$ is the applied transverse force per unit length.
